```python

import os, io
import torch
from diffusers import StableDiffusionPipeline
from PIL import Image

OUT_DIR = "outputs/baseline"
os.makedirs(OUT_DIR, exist_ok=True)

MODEL_ID = "runwayml/stable-diffusion-v1-5"  # baseline (no LoRA)
device = "cuda" if torch.cuda.is_available() else "cpu"
dtype = torch.float16 if device=="cuda" else torch.float32

pipe = StableDiffusionPipeline.from_pretrained(MODEL_ID, torch_dtype=dtype)
pipe = pipe.to(device)
pipe.enable_attention_slicing()

def generate_and_save(prompt, out_path, seed=None, steps=40, guidance=7.5):
    g = None
    if seed is not None:
        g = torch.Generator(device=device).manual_seed(seed)
    image = pipe(prompt, num_inference_steps=steps, guidance_scale=guidance, generator=g, num_images_per_prompt=1).images[0]
    image.save(out_path)
    return out_path

# 1) generate 20 images 
N = 20
base_prompt = "photo of chikankari embroidery"  
for i in range(N):
    out_p = os.path.join(OUT_DIR, f"baseline_20_{i+1:02d}.png")
    seed = 1000 + i
    generate_and_save(base_prompt, out_p, seed=seed)
    if device == "cuda":
        torch.cuda.empty_cache()

# 2) generate images for the 4 sample prompts (one or more images per prompt)
sample_prompts = [
    "a photo of chikankari embroidery",
    "floral paisley vines embroidery",
    "delicate lotus buti embroidery",
    "geometric chikankari border"
]
SAMPLE_DIR = os.path.join(OUT_DIR, "sample_prompts")
os.makedirs(SAMPLE_DIR, exist_ok=True)
for idx, p in enumerate(sample_prompts, start=1):
    prompt_dir = os.path.join(SAMPLE_DIR, f"prompt_{idx}")
    os.makedirs(prompt_dir, exist_ok=True)
    out_p = os.path.join(prompt_dir, f"baseline_sample_{idx}.png")
    generate_and_save(p, out_p, seed=2000+idx)
    if device == "cuda":
        torch.cuda.empty_cache()

print("Done: baseline images saved to", OUT_DIR)
```

```python

import os, glob, csv
import torch
import lpips
from PIL import Image
import torchvision.transforms as T

auth_dir = "data/authentic_for_eval"   # put the 20 authentic images here
gen_dir  = "outputs/baseline"
out_csv  = "baseline_lpips_20.csv"

# get sorted lists (ensure pairing order matches how fine-tuned eval was done)
auth_paths = sorted(glob.glob(os.path.join(auth_dir, "*.jpg")))[:20]
gen_paths  = sorted(glob.glob(os.path.join(gen_dir, "baseline_20_*.png")))

assert len(auth_paths) == len(gen_paths), "Mismatch counts; ensure pairing"

# LPIPS model
loss_fn = lpips.LPIPS(net='alex')
device = 'cuda' if torch.cuda.is_available() else 'cpu'
loss_fn = loss_fn.to(device)

transform = T.Compose([
    T.Resize((256,256)),
    T.ToTensor(),
    T.Normalize((0.5,0.5,0.5),(0.5,0.5,0.5))
])

rows = [("auth_image","gen_image","lpips")]
for a,g in zip(auth_paths, gen_paths):
    img_a = transform(Image.open(a).convert("RGB")).unsqueeze(0).to(device)
    img_g = transform(Image.open(g).convert("RGB")).unsqueeze(0).to(device)
    with torch.no_grad():
        dist = loss_fn(img_a, img_g).item()
    rows.append((os.path.basename(a), os.path.basename(g), dist))
    print(a, dist)

# write CSV
with open(out_csv, "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)
print("Saved LPIPS csv:", out_csv)
```

```python
from transformers import CLIPProcessor, CLIPModel
import torch
from PIL import Image
import os
import pandas as pd

device = "cuda" if torch.cuda.is_available() else "cpu"

# Load Hugging Face CLIP
model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32").to(device)
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")

prompt = "floral paisley embroidery design"
image_dir = "outputs/baseline_resized/"
image_paths = [os.path.join(image_dir, f) for f in os.listdir(image_dir) if f.endswith((".png", ".jpg", ".jpeg"))]

scores = []
for p in image_paths:
    image = Image.open(p).convert("RGB")
    inputs = processor(text=[prompt], images=image, return_tensors="pt", padding=True).to(device)

    with torch.no_grad():
        outputs = model(**inputs)
        logits_per_image = outputs.logits_per_image  # similarity scores
        score = logits_per_image.item()
        scores.append((os.path.basename(p), score))

df = pd.DataFrame(scores, columns=["image", "clipscore"])
df.to_csv("baseline_clipscores.csv", index=False)
print("✅ Saved CLIP scores to baseline_clipscores.csv")
```
